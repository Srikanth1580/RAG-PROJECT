```python
print("Using Docx2txtLoader:")

try:
    docx_loader = Docx2txtLoader("data/word_files/proposal.docx")
    docx_docs = docx_loader.load()
    print(f"Docx2txtLoader extracted {len(docx_docs)} document(s).")
    print(f"First document content (truncated): {docx_docs[0].page_content[:200]}...")
    print(f"metadata: {docx_docs[0].metadata}")
except Exception as e:
    print(f"Error using Docx2txtLoader: {e}")
```

```python
print("UnstructuredWordDocumentLoader:")

try:
    docx_loader = UnstructuredWordDocumentLoader("data/word_files/proposal.docx", mode="elements")
    docx_docs = docx_loader.load()
    print(f"UnstructuredWordDocumentLoader extracted {len(docx_docs)} document(s).")

    for i, doc in enumerate(docx_docs[:10]):  # Print first 10 documents
        print(f"\nDocument {i+1} content (truncated): {doc.page_content[:200]}...")
        print(f"metadata: {doc.metadata}")

except Exception as e:
    print(f"Error using UnstructuredWordDocumentLoader: {e}")
```
