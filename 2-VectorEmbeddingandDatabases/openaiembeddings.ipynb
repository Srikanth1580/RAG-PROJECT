```python
from langchain_openai import OpenAIEmbeddings
embedding_model = OpenAIEmbeddings(model="text-embedding-3-small")
embedding_model
```

```python
## single text embedding

single_text = "sri rama temple is located in ayodhya in uttar pradesh state of india"

single_embedding = embedding_model.embed_query(single_text)

print(f"Embedding for the text: {single_embedding}")
print(f"Embedding dimension: {len(single_embedding)}")
```
